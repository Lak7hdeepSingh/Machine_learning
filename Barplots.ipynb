```python
import matplotlib.pyplot as plt
```

```python
years = [2008,2009,2010,2011,2012]
oscar_revenue = [1005,170,427,133,232]

plt.bar(years,oscar_revenue)
plt.title("Oscar movies Revenue in each Year")
plt.xlabel("Years")
plt.ylabel("Revenue in Million $")
```

```python
years = [2008,2009,2010,2011,2012]
oscar_revenue = [1005,170,427,133,232]

plt.bar(years,oscar_revenue,color="purple")
plt.title("Oscar movies Revenue in each Year")
plt.xlabel("Years")
plt.ylabel("Revenue in Million $")

for i in range(len(years)):
    plt.text(years[i],oscar_revenue[i]+5,str(oscar_revenue[i]),ha="center",color="blue")

plt.ylim(0,max(oscar_revenue)+100)
```

```python
age = [22,25,30,35,40,45,50,55,60,65]
blood_pressure = [110,115,120,122,125,130,135,123,145,150]


plt.scatter(age,blood_pressure)
plt.title("AGE vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")

plt.grid()
```

```python
age = [22,25,30,35,40,45,50,55,60,65]
blood_pressure = [110,115,120,122,125,130,135,123,145,150]
colors = [
    "red",
    "purple",
    "blue",
    "green",
    "yellow",
    "orange",
    "pink",
    "brown",
    "black",
    "red"
]


plt.scatter(age,blood_pressure,color=colors)
plt.title("AGE vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")

plt.grid()
```

```python
age = [22,25,30,35,40,45,50,55,60,65]
bp = [110,115,120,122,125,130,135,123,145,150]
colors = ["green" if x<135 else "red" for x in bp]


plt.scatter(age,bp,color=colors,s=bp,marker="x")
plt.title("AGE vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")

plt.grid()
```
